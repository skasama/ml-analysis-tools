```python
from pathlib import Path

import numpy as np

from analysis_tools.io import (
    list_hdf5_datasets,
    load_array_pair,
    load_hdf5,
    load_npy,
    load_npz,
)
```

```python
PREDICTIONS_PATH = Path("path/to/outputs/preds.npy")
TARGETS_PATH = Path("path/to/outputs/targets.npy")

NPY_PATH = Path("path/to/array.npy")
NPZ_PATH = Path("path/to/data.npz")
HDF5_PATH = Path("path/to/data.hy")

NPZ_KEYS = [
    "train_idxs",
    "val_idxs",
    "test_idxs",
]

HDF5_KEYS = [
    "labels",
    "positions",
    "directions",
    "energies",
]
```

```python
def print_array_summary(name: str, array: np.ndarray) -> None:
    """Print a compact summary of an array."""
    array = np.asarray(array)

    print(f"\n{name}")
    print(f"  Shape: {array.shape}")
    print(f"  Dtype: {array.dtype}")

    if array.ndim > 0:
        print(f"  Events: {len(array):,}")

    if np.issubdtype(array.dtype, np.number):
        finite = np.isfinite(array)
        print(
            "  Finite values: "
            f"{np.count_nonzero(finite):,} / {array.size:,}"
        )

        if np.any(finite):
            finite_values = array[finite]
            print(
                "  Range: "
                f"{np.min(finite_values):.6g} to "
                f"{np.max(finite_values):.6g}"
            )

    if array.ndim > 0:
        print("  First entries:")
        print(array[: min(3, len(array))])
    else:
        print(f"  Value: {array}")
```

```python
# predictions, targets
predictions, targets = load_array_pair(
    PREDICTIONS_PATH,
    TARGETS_PATH,
    first_name="predictions",
    second_name="targets",
)

print_array_summary("Predictions", predictions)
print_array_summary("Targets", targets)
```

```python
# .npz file
array = load_npy(NPY_PATH)
print_array_summary(NPY_PATH.name, array)
```

```python
# inspect .npz file
with np.load(NPZ_PATH, allow_pickle=False) as archive:
    print("Available NPZ keys:")
    for key in archive.files:
        print(f"  {key}")
arrays = load_npz(NPZ_PATH, keys=NPZ_KEYS)

for key, array in arrays.items():
    print_array_summary(key, array)
```

```python
# inspect .hdf5 file
datasets = list_hdf5_datasets(HDF5_PATH)

print(f"HDF5 file: {HDF5_PATH}")
print(f"Datasets: {len(datasets)}")
for key in datasets:
    print(f"  {key}")

arrays = load_hdf5(HDF5_PATH, keys=HDF5_KEYS)

for key, array in arrays.items():
    print_array_summary(key, array)
```
